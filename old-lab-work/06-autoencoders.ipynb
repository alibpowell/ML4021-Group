Then answer briefly:

1. What is the main idea behind an autoencoder?  

An autoencoder is a type of algorithm with the primary purpose of learning an "informative" representation of the data that can be used for different applications by learning to reconstruct a set of input observations well enough. 

2. What is the role of the bottleneck layer?  

Compresses the input into a smaller representation, forcing the autoencoder to learn the most important features instead of just copying the input - if the bottleneck is too small it loses important information, making reconstruction less accurate. 

3. Discuss, in your own words, the applications of autoencoders mentioned in the paper.  

One application of autoencoders is dimensionality reduction, where the encoder compresses the original data into a smaller set of useful features. These features can then be used for tasks like classification and clustering. Autoencoders can also be used for anomaly detection by training on normal data and identifying unusual observations that have higher reconstruction errors. Another application is denoising, where the model learns to reconstruct cleaner versions of noisy inputs.

4. What differences would you expect between an autoencoder’s representation and that of PCA if applied as a dimensionality reduction technique?

The first difference is that PCA is a linear transformation whereas autoencoders use nonlinear neural networks through activation functions. PCA also maximizes variance through orthogonal projections while autoencoders minimize reconstruction error.

## 3. Autoencoder

In this section, you will build and train an **autoencoder** to reconstruct the provided data.

Your model will include three layers in both the encoder and decoder:

- **Encoder:** 784 → 128 → 64 → 2. This last layer here is the **bottleneck**.
- **Decoder:** 2 → 64 → 128 → 784  

### 3.1 Implementation (30 points)

Follow the skeleton provided below and fill in the missing parts (`#TODO`). 

Follow these restrictions:

- Each **hidden** layer should include a linear transformation, ReLU activation function, dropout with `p=0.2` and batch normalization steps. Determine the appropriate order for these components based on what we discussed in class. The bottleneck layer is the exception; see next point.
- For the bottleneck, use only a linear transformation, i.e., no activation, batch normalization, or dropout.
- The output layer should have a sigmoid activation function.

```python
import torch.nn as nn

class Autoencoder(nn.Module):
    def __init__(self):
        super().__init__()
        
        # ENCODING PART: 784 -> 128 -> 64 -> 2
        encoder_list = [
            nn.Linear(784, 128), nn.BatchNorm1d(128), nn.ReLU(), nn.Dropout(0.2),
            nn.Linear(128, 64), nn.BatchNorm1d(64), nn.ReLU(), nn.Dropout(0.2),
            nn.Linear(64, 2),
        ]
        self.encoder_list = nn.ModuleList(encoder_list)
        
        # DECODING PART: 2 -> 64 -> 128 -> 784
        decoder_list = [
            nn.Linear(2, 64), nn.BatchNorm1d(64), nn.ReLU(), nn.Dropout(0.2),
            nn.Linear(64, 128), nn.BatchNorm1d(128), nn.ReLU(), nn.Dropout(0.2),
            nn.Linear(128, 784), nn.Sigmoid(),
        ]
        self.decoder_list = nn.ModuleList(decoder_list)
        
    def forward(self, X):
        output = X
        for layer in self.encoder_list:
            output = layer(output)
        for layer in self.decoder_list:
            output = layer(output)
        return output
```

### 3.2 Training (15 points)

Follow the skeleton provided below and fill in the missing parts to train your model (`#TODO`). As you can see, training includes EarlyStopping, which I give you already for your convenience :-). - The Sigmoid activation function keeps the output values between 0 and 1, which matches the range of our normalized MNIST images. This allows the autoencoder to reconstruct pixel values within the same range as the original images.

```python
model = Autoencoder()  

criterion = nn.MSELoss()  

# Use learning rate of 0.001 and an L2 regularization via weight_decay of 1e-5
optimizer = torch.optim.Adam(model.parameters(), lr=0.001, weight_decay=1e-5)  
```

```python
class EarlyStopping:
    def __init__(self, patience=10, min_delta=0.0):
        self.patience = patience
        self.min_delta = min_delta
        self.counter = 0
        self.best_loss = None
        self.early_stop = False
        self.best_model_state = None

    def __call__(self, tuning_loss, model):
        if self.best_loss is None or tuning_loss < self.best_loss - self.min_delta:
            self.best_loss = tuning_loss
            self.counter = 0
            self.best_model_state = {k: v.clone() for k, v in model.state_dict().items()}
        else:
            self.counter += 1
            if self.counter >= self.patience:
                self.early_stop = True

    def restore_best_weights(self, model):
        if self.best_model_state is not None:
            model.load_state_dict(self.best_model_state)


def train_autoencoder(model, fitting_loader, tuning_loader, n_epochs=100, patience=5):
    # Initialize the early stopping object
    early_stopping = EarlyStopping(patience=patience)
    history = {"fitting": [], "tuning": []}

    for epoch in range(n_epochs):
        model.train()
        fitting_cost = 0.0

        for X_batch, _ in fitting_loader:
            # --- Implement the standard training steps ---
            
            outputs = model(X_batch)
            cost = criterion(outputs, X_batch)
            cost.backward()
            optimizer.step()
            optimizer.zero_grad()
            
            fitting_cost += cost.item() * X_batch.size(0)

        fitting_cost /= len(fitting_loader.dataset)

        # --- Tuning phase ---
        model.eval()
        tuning_cost = 0.0
        with torch.no_grad():
            for X_batch, _ in tuning_loader:
                outputs = model(X_batch)
                cost = criterion(outputs, X_batch)
                tuning_cost += cost.item() * X_batch.size(0)

        tuning_cost /= len(tuning_loader.dataset)

        # --- Save metrics ---
        history["fitting"].append(fitting_cost)
        history["tuning"].append(tuning_cost)

        print(f"Epoch [{epoch+1}/{n_epochs}]  Fitting Cost: {fitting_cost:.5f}  Tuning Cost: {tuning_cost:.5f}")

        # --- Early stopping check ---
        early_stopping(tuning_cost, model)
        if early_stopping.early_stop:
            print(f"Stopping early at epoch {epoch+1}")
            early_stopping.restore_best_weights(model)
            return model, history

    # Also restore the best weights when the epoch limit is reached.
    early_stopping.restore_best_weights(model)
    return model, history
```

## 4. Visualize the Learned Representations (15 points)

Here you will explore the latent representations learned by your autoencoder.

1. Complete `extract_latent_features` to extract the latent (bottleneck) features and their corresponding labels, given a trained model.  
   Apply this function to the reporting set using the provided code. This should give you the transformed reporting set with the dimensions of the bottleneck (2).
2. Plot the reduced 2D data as a scatterplot, coloring each point by its corresponding digit label.


```python
def extract_latent_features(model, dataloader): 
    model.eval()
    latents = []
    labels = []

    with torch.no_grad():
        for X_batch, y_batch in dataloader:
            
            encoded = X_batch
            for layer in model.encoder_list:
                encoded = layer(encoded)
            latents.append(encoded.numpy())
            labels.append(y_batch.numpy())

    # Concatenate all batches
    latents = np.concatenate(latents, axis=0)
    labels = np.concatenate(labels, axis=0)
    return latents, labels
```

Once the missing parts above have been filled, uncomment the piece of code below and execute it to extract the latent features

```python
latents_autoencoder, labels = extract_latent_features(model, reporting_loader)

print("Latent features shape:", latents_autoencoder.shape)
print("Labels shape:", labels.shape)
```

```python
# scatterplot

plt.figure(figsize=(9, 6))

scatter = plt.scatter(
    latents_autoencoder[:, 0],
    latents_autoencoder[:, 1],
    c=labels,
    cmap="tab10",
    vmin=-0.5,
    vmax=9.5,
    s=8,
    alpha=0.6
)

plt.colorbar(scatter, ticks=range(10), label="Actual digit")
plt.xlabel("Latent feature 1")
plt.ylabel("Latent feature 2")
plt.title("Autoencoder- 2D Representation of Reporting Digits")
plt.tight_layout()
plt.show()
```

Here you will compare the latent representations learned by your autoencoder with those obtained using a PCA.

1. Apply a PCA to the reporting set (without using the autoencoder) to reduce the data to the same dimensionality as your bottleneck (2 dimensions). Obviously, first use the rest of the data to learn these two components.

```python
# 1
from sklearn.decomposition import PCA
import numpy as np

# learn the components without using the reporting set
X_pca_fit = np.concatenate([X_fitting, X_tuning], axis=0)

pca = PCA(n_components=2, random_state=42)
pca.fit(X_pca_fit)

# represent each reporting image with two numbers
latents_pca = pca.transform(X_reporting)

print("PCA reporting shape:", latents_pca.shape)

# 2

import matplotlib.pyplot as plt

plt.figure(figsize=(9, 6))

scatter = plt.scatter(
    latents_pca[:, 0],
    latents_pca[:, 1],
    c=y_reporting,
    cmap="tab10",
    vmin=-0.5,
    vmax=9.5,
    s=8,
    alpha=0.6
)

plt.colorbar(scatter, ticks=range(10), label="Actual digit")
plt.xlabel("Principal component 1")
plt.ylabel("Principal component 2")
plt.title("PCA: 2D Representation of Reporting Digits")
plt.tight_layout()
plt.show()
```

Comparing the Autoencoder and PCA - The autoencoder’s plot shows more distinct branches for some digits, such as 0, 1, and 6, whereas the PCA plot shows broader regions with substantial overlap between digit labels. However, both plots still contain overlapping groups, so neither method completely separates the digits in two dimensions. The autoencoder can learn nonlinear relationships in the images, while PCA finds linear directions that capture the greatest variation in pixel values. This may help explain the different structures in the plots. Overall, the autoencoder appears to produce a more visually distinct representation for some digits, but this doesn’t prove that it performs better for every digit or task.
